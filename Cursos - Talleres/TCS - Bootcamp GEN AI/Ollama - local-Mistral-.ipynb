# 1 Usar una función personalizada para reutilizar el chat

Si planeas realizar múltiples interacciones, encapsular la lógica en una función te permite reutilizarla fácilmente:

```python
import ollama

def chat_with_model(model, user_message):
    response = ollama.chat(model=model, messages=[{'role': 'user', 'content': user_message}])
    return response['message']['content']

# Uso
user_input = "¿Cuándo sembrar papa en las montañas de Colombia y a qué altitud hacerlo para que sea más productivo?"
response = chat_with_model('mistral', user_input)
print(response)
```

# 2. Usar un historial de mensajes para chats continuos
Para mantener un contexto continuo, puedes construir un historial de mensajes dinámico:

```python
import ollama

# Historial de mensajes
chat_history = []

def chat_with_history(model, user_message):
    chat_history.append({'role': 'user', 'content': user_message})
    response = ollama.chat(model=model, messages=chat_history)
    chat_history.append({'role': 'assistant', 'content': response['message']['content']})
    return response['message']['content']

# Interacción continua
user_input = "¿Cuándo sembrar papa en las montañas de Colombia y a qué altitud hacerlo?"
response = chat_with_history('mistral', user_input)
print(response)

# Añadir otra pregunta
user_input = "¿Qué fertilizantes son recomendables?"
response = chat_with_history('mistral', user_input)
print(response)
```

# 3. Usar un generador para respuestas interactivas
Si esperas múltiples respuestas interactivas, puedes implementar un generador:

```python
import ollama

def interactive_chat(model):
    chat_history = []
    while True:
        user_message = input("Tú: ")
        if user_message.lower() in ['salir', 'exit']:
            print("Chat terminado.")
            break
        chat_history.append({'role': 'user', 'content': user_message})
        response = ollama.chat(model=model, messages=chat_history)
        print(f"Modelo: {response['message']['content']}")
        chat_history.append({'role': 'assistant', 'content': response['message']['content']})

# Inicia el chat interactivo
interactive_chat('mistral')
```

```python
from fastapi import FastAPI
import ollama

app = FastAPI()

@app.post("/chat")
def chat_with_model(data: dict):
    user_message = data.get('message', '')
    response = ollama.chat(model='mistral', messages=[{'role': 'user', 'content': user_message}])
    return {"response": response['message']['content']}

# Ejecuta este script y accede a la API en /chat
```

# 5. Ejecutar múltiples modelos con un pipeline
Si trabajas con varios modelos, puedes implementar un pipeline para distribuir las tareas entre ellos:

```python
import ollama

def chat_pipeline(user_message):
    models = ['mistral', 'llama2','llama3']  # Lista de modelos
    responses = []
    for model in models:
        response = ollama.chat(model=model, messages=[{'role': 'user', 'content': user_message}])
        responses.append(f"{model}: {response['message']['content']}")
        responses.append("-------------------")
    return responses

# Uso
user_input = "¿Cuándo sembrar papa en las montañas de Colombia?"
responses = chat_pipeline(user_input)
for res in responses:
    print(res)
```

# 6. Guardar y cargar conversaciones
Si deseas guardar el historial para futuras sesiones:

```python
import ollama
import json

def save_chat_history(file_name, chat_history):
    with open(file_name, 'w') as file:
        json.dump(chat_history, file)

def load_chat_history(file_name):
    with open(file_name, 'r') as file:
        return json.load(file)

# Ejemplo
chat_history = [{'role': 'user', 'content': '¿Cuándo sembrar papa?'}]
response = ollama.chat(model='mistral', messages=chat_history)
chat_history.append({'role': 'assistant', 'content': response['message']['content']})
save_chat_history('chat_history.json', chat_history)
print("Historial guardado.")
```

# Determinando el tiempo de respuesta de cada Modelo


```python
import ollama
import time  # Importar módulo para medir el tiempo

def chat_pipeline(user_message):
    models = ['mistral', 'llama2', 'llama3','llama3.1','llama3.2-vision']  # Lista de modelos
    responses = []
    for model in models:
        start_time = time.time()  # Inicio del contador
        response = ollama.chat(model=model, messages=[{'role': 'user', 'content': user_message}])
        end_time = time.time()  # Fin del contador
        
        elapsed_time = end_time - start_time  # Tiempo transcurrido
        responses.append(f"{model}: {response['message']['content']}")
        responses.append(f"Tiempo de respuesta: {elapsed_time:.2f} segundos")
        responses.append("-------------------")
    return responses

# Uso
user_input = "¿Cuándo sembrar papa en las montañas de Colombia?"
responses = chat_pipeline(user_input)
for res in responses:
    print(res)
```
